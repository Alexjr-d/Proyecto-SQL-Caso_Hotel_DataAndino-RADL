INNER JOIN (Básico): Une tb_tickets con tb_productos usando ID_Producto. Muestra el número de Ticket, la Fecha, el Nombre_producto y la Cantidad.

```python
%sql
SELECT
    P.*,
    S.*
FROM cursodataacademy.proyecto_kiosko.data_tickets P
JOIN cursodataacademy.proyecto_kiosko.data_productos S
ON P.ID_Producto = S.ID_Producto
LIMIT 10;
```

FULL OUTER JOIN: Haz un cruce completo entre productos y tickets. Esto permitirá a auditoría detectar si existen tickets que tengan un ID_Producto que ya no existe en el catálogo, o productos que no tengan tickets.

```python
%sql
-- Que productos no estan dentro de la tabla de productos, pero que si se encuentrann en la tabla tickets
SELECT p.Nombre_producto, t.Ticket, t.Fecha
FROM cursodataacademy.proyecto_kiosko.data_productos p
FULL OUTER JOIN cursodataacademy.proyecto_kiosko.data_tickets t 
ON p.ID_Producto = t.ID_Producto
WHERE P.ID_Producto IS NULL;
```

```python
%sql
-- Que productos no se estan vendiendo
SELECT p.Nombre_producto, t.Ticket, t.Fecha
FROM cursodataacademy.proyecto_kiosko.data_productos p
FULL OUTER JOIN cursodataacademy.proyecto_kiosko.data_tickets t 
ON p.ID_Producto = t.ID_Producto
WHERE t.ID_Producto IS NULL;
     
```
